```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

labels = [
    "Natural Language Processing",
    "Computer Vision",
    "Machine Learning",
    "Deep Learning",
    "Healthcare AI",
    "Robotics",
    "Cybersecurity",
    "Data Mining",
    "Reinforcement Learning"
]

def classify_topic(text):
    result = classifier(text, candidate_labels=labels)
    return result["labels"][0], result["scores"][0]

print("Topic Classification Loaded")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def extract_entities(text):
    doc = nlp(text)
    return [(ent.text, ent.label_) for ent in doc.ents]

print("NER Loaded Successfully")
```

```python
text = """
Deep learning models are widely used for detecting lung cancer from medical images using convolutional neural networks.
"""

topic, confidence = classify_topic(text)

print("Predicted Topic:", topic)
print("Confidence:", round(confidence * 100, 2), "%")
```

```python
text = """
Researchers from Stanford University developed an AI model using TensorFlow to analyze MRI scans at Google Health.
"""

entities = extract_entities(text)

print("Named Entities:")
for entity, label in entities:
    print(f"{entity} --> {label}")
```
